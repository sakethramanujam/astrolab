# 🌌 Module 1: Anatomy of a FITS File

Welcome to your first lab! In professional astronomy, we don't use JPEGs. We use **FITS** (Flexible Image Transport System) files. Unlike a standard image, a FITS file contains:
- **The Data**: Raw pixel values representing the number of photons hitting the sensor.
- **The Header**: Metadata about the telescope, the observer, and the precise coordinates in the sky.

---

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.utils.data import download_file
from astropy.visualization import astropy_mpl_style, LogStretch, ImageNormalize

# Set the plotting style to look 'astronomical'
plt.style.use(astropy_mpl_style)
```

## 1. Fetching Real Space Data
We are going to download an image of **M13 (The Great Globular Cluster in Hercules)**. This image contains thousands of stars, making it perfect for testing our compute power later.

```python
print("🛰️ Downloading M13 FITS image from the archive...")
image_url = 'https://archive.stsci.edu/pub/vosamp/horsehead.fits' # Placeholder for M13/Horsehead
fits_file = download_file(image_url, cache=True)
print("✅ Download complete!")
```

## 2. Reading the 'Label' (The Header)
Let's see what the telescope recorded about this observation.

```python
with fits.open(fits_file) as hdul:
    # HDU = Header Data Unit
    header = hdul[0].header
    data = hdul[0].data

print(f"🔭 Instrument: {header.get('INSTRUME', 'Unknown')}")
print(f"⏰ Exposure: {header.get('EXPTIME', 'Unknown')} seconds")
print(f"📏 Dimensions: {data.shape} pixels")
```

## 3. Visualizing the Invisible
Raw space data is often very dark because stars are much brighter than the background. If we use a normal linear scale, we only see the brightest stars. We use **Logarithmic Scaling** to see everything.

```python
# Normalize the image using a Log scale to reveal faint details
norm = ImageNormalize(data, stretch=LogStretch())

plt.figure(figsize=(12, 10))
plt.imshow(data, cmap='inferno', origin='lower', norm=norm)
plt.colorbar(label='Photon Counts')
plt.title('Deep Space Visualization: M13 Region')
plt.xlabel('RA (Pixels)')
plt.ylabel('Dec (Pixels)')
plt.show()
```

# 🐝 Module 2: Scaling to the Stars with Dask

In Module 1, we processed a single image. But what if you have 10,000 images from a full night of observing? 

This is where your **NUC**, **Windows WSL**, and **Raspberry Pi** come in. We will use **Dask Distributed** to turn your home network into a mini-supercomputer cluster.

---

```python
import time
import numpy as np
from dask.distributed import Client, progress
import dask

# 1. Connect to your Cluster
# If you have started 'dask-scheduler' on your NUC, put its IP here.
# Otherwise, this will create a local cluster using all cores on THIS machine.
try:
    client = Client('127.0.0.1:8786', timeout="2s") 
    print("🔗 Connected to your NETWORK cluster!")
except:
    client = Client(n_workers=4) 
    print("🏠 No network cluster found. Started a LOCAL cluster instead.")

client
```

## 2. Defining a 'Lazy' Astronomical Task
In distributed computing, we use **Lazy Evaluation**. Instead of running the code immediately, we build a 'Task Graph' of what needs to be done.

```python
@dask.delayed
def calibrate_image(image_id):
    """Simulates subtracting dark frames and flat-fielding an image."""
    # Simulate some heavy math (1 second per image)
    time.sleep(1) 
    
    # Generate a dummy reduced image result
    result_mean = np.random.random() * 100
    return f"Image {image_id} processed. Mean: {result_mean:.2f}"

print("✅ Function defined as a 'Delayed' task.")
```

## 3. Parallel Execution across the Fleet
Let's process 20 images. If you have 4 workers (e.g., your NUC and WSL), this should take about 5 seconds instead of 20!

```python
num_images = 20
print(f"🚀 Sending {num_images} tasks to your cluster nodes...")

# Build the list of tasks
tasks = [calibrate_image(i) for i in range(num_images)]

# Execute them in parallel!
start_time = time.time()
results = dask.compute(*tasks)
end_time = time.time()

print(f"\n✅ Finished processing in {end_time - start_time:.2f} seconds.")
for r in results[:5]: # Show first 5 results
    print(f"   - {r}")
```

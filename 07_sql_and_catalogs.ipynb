# 🗄️ Module 7: SQL and Big Data Catalogs

In astronomy, we often don't have the luxury of downloading every image. Instead, we query **Databases (SQL)** to filter for exactly what we need (e.g., "Give me all blue stars within 5 arcseconds of this galaxy").

---

```python
from astroquery.sdss import SDSS
import pandas as pd

# 1. Complex SQL Query to SDSS (Sloan Digital Sky Survey)
# Filtering for galaxies with high redshift (far away)
query = """
SELECT TOP 10
  objid, ra, dec, u, g, r, i, z
FROM Galaxy
WHERE 
  u BETWEEN 0 AND 19
  AND g BETWEEN 0 AND 19
"""

print("📡 Executing SQL query against SDSS Database...")
result = SDSS.query_sql(query)

if result:
    df = result.to_pandas()
    print("✅ Query Success!")
    display(df.head())
else:
    print("❌ No results found or query error.")
```
